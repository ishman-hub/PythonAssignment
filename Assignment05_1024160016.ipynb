```python
#Ques1
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print("Original Array:", arr)

add_res = arr + 2
print("a) Add 2 to all elements:", add_res)

mul_res = arr * 3
print("b) Multiply 3 with all elements:", mul_res)

div_res = arr / 2
print("c) Divide every element by 2:", div_res)
```

```python
# Question 2
# a) Reverse the NumPy array
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print(reversed_arr)

# b) Find the most frequent value and their indice(s)
# i.
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
vals_x, counts_x = np.unique(x, return_counts=True)
most_freq_x = vals_x[counts_x == np.max(counts_x)]
for val in most_freq_x:
    print(val, np.where(x == val)[0])

# ii.
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
vals_y, counts_y = np.unique(y, return_counts=True)
most_freq_y = vals_y[counts_y == np.max(counts_y)]
for val in most_freq_y:
    print(val, np.where(y == val)[0])
```

```python
# Question 4
Naman = np.linspace(10, 100, 25)

print(Naman.ndim)
print(Naman.shape)
print(Naman.size)
print(Naman.dtype)
print(Naman.nbytes)

transposed_reshape = Naman.reshape(-1, 1)
print(transposed_reshape)

# Can we do the same with T attribute?
# No, .T on a 1D array returns the unchanged 1D array because it has only one axis.
print(Naman.T)
```

```python
# Question 5
values = [10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]
ucs420_Naman = np.array(values).reshape(3, 4)
print(ucs420_Naman)

print(np.mean(ucs420_Naman))
print(np.median(ucs420_Naman))
print(np.max(ucs420_Naman))
print(np.min(ucs420_Naman))
print(np.unique(ucs420_Naman))

reshaped_ucs420_Naman = ucs420_Naman.reshape(4, 3)
print(reshaped_ucs420_Naman)

resized_ucs420_Naman = np.resize(ucs420_Naman, (2, 3))
print(resized_ucs420_Naman)
```
