```python
# Question 1
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2°C to every reading using vectorization
temp_corrected = temperature + 2
print(temp_corrected)

# b. Convert all temperatures from Celsius to Fahrenheit using F = (9/5)C + 32
temp_f = (9 / 5) * temp_corrected + 32
print(temp_f)

# c. Identify all readings greater than 32°C using Boolean indexing
above_32 = temp_corrected[temp_corrected > 32]
print(above_32)

# d. Count how many readings exceed 32°C
count_above_32 = np.sum(temp_corrected > 32)
print(count_above_32)

# e. Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value
# Vectorization delegates element-wise computations to pre-compiled C-level code, avoiding Python bytecode overhead and dynamic type-checking.
# Boolean indexing operates directly on contiguous memory blocks via C arrays, eliminating explicit loop counter tracking and list lookups.
```

```python
# Question 2
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Calculate the total steps recorded
print(np.sum(steps))

# b. Calculate the mean number of steps
print(np.mean(steps))

# c. Find the maximum and minimum values
print(np.max(steps), np.min(steps))

# d. Calculate the total steps for each day using axis=0
print(np.sum(steps, axis=0))

# e. Calculate the total steps for each user using axis=1
print(np.sum(steps, axis=1))

# f. Find the user/day position containing the maximum number of steps using argmax()
flat_idx = np.argmax(steps)
user_idx, day_idx = np.unravel_index(flat_idx, steps.shape)
print((user_idx, day_idx))
```

```python
# Question 4
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

# a. Display the shape and dimensions of X
print(X.shape)
print(X.ndim)

# b. Calculate: X.T and explain what the transpose represents
# Transpose switches rows and columns: each row now represents a feature across users, and each column represents a user.
print(X.T)

# c. Calculate the matrix product: X.T @ X
xt_x = X.T @ X
print(xt_x)

# d. Calculate the matrix inverse using: np.linalg.inv() where appropriate
xt_x_inv = np.linalg.inv(xt_x)
print(xt_x_inv)

# e. Implement the Ordinary Least Squares equation: beta = (X^T X)^(-1) X^T y using NumPy
beta = xt_x_inv @ X.T @ y
print(beta)

# f. Explain what the resulting coefficients represent in the context of the three user features
# Beta coefficients quantify the linear impact of Sleep hours, Activity level, and Stress level on the assistance score.
# A positive coefficient indicates higher assistance needed as that feature increases; a negative coefficient indicates lower assistance needed.

# g. A new user has: new_user = np.array([5, 40, 7]) Use your calculated coefficients to obtain a predicted assistance score
new_user = np.array([5, 40, 7])
pred_score = new_user @ beta
print(pred_score)
```
